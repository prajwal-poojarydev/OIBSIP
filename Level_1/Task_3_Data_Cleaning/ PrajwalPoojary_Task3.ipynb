### Initial Data Quality Findings

The initial inspection identified the following issues:

- Missing values in `email`, `signup_date`, and `spend`.
- Duplicate records and repeated customer IDs requiring further investigation.
- Inconsistent name capitalisation, country labels, customer segments, and active-status values.
- Mixed date formats and invalid date placeholders such as `none`, `-`, and `-999`.
- Spending values stored as text, including currency symbols, missing-value placeholders, and negative sentinel values.
- All columns currently use the `object` data type, so appropriate data types must be established during cleaning.

These issues will be investigated before applying cleaning operations. Repeated customer IDs will not automatically be treated as duplicate rows because a customer may have multiple records.


```python
print("Original data types:")
print(df.dtypes)

print("\nUnique active-status values:")
print(sorted(df["is_active"].dropna().unique()))

print("\nUnique segment values:")
print(sorted(df["segment"].dropna().unique()))

print("\nSample spending values:")
print(df["spend"].dropna().sample(min(20, df["spend"].notna().sum()), random_state=42).tolist())
```

```python
spend_text = df["spend"].astype("string").str.strip()

print("Spending placeholders:")
print(
    spend_text[
        spend_text.str.lower().isin(["", "-", "none", "null", "nan", "-999"])
    ].value_counts(dropna=False)
)

print("\nExamples of spending values:")
print(spend_text.dropna().unique()[:30])
```

```python
spend_clean = (
    df["spend"]
    .astype("string")
    .str.strip()
    .str.replace(r"(?i)^USD\s*", "", regex=True)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

spend_clean = spend_clean.replace(
    r"(?i)^\s*(?:-|none|null|nan|-999)?\s*$",
    pd.NA,
    regex=True
)

spend_numeric = pd.to_numeric(spend_clean, errors="coerce")

print("Valid numeric spending values:", spend_numeric.notna().sum())
print("Unusable or missing values:", spend_numeric.isna().sum())

print("\nSpending summary:")
print(spend_numeric.describe())

print("\nLargest 15 spending values:")
print(spend_numeric.nlargest(15).to_list())
```

```python
Q1 = spend_numeric.quantile(0.25)
Q3 = spend_numeric.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = spend_numeric[
    (spend_numeric < lower_bound) |
    (spend_numeric > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))

print("\nLargest detected outliers:")
print(outliers.nlargest(10).to_list())
```

```python
outlier_rows = df.loc[
    spend_numeric > upper_bound,
    ["customer_id", "full_name", "spend"]
].copy()

outlier_rows["numeric_spend"] = spend_numeric.loc[outlier_rows.index]

outlier_rows.sort_values(
    "numeric_spend", ascending=False
).head(20)
```

```python
duplicate_rows = df[df.duplicated(keep=False)].sort_values("customer_id")

duplicate_rows
```

```python
df_clean = df.drop_duplicates().copy()

print("Rows before cleaning:", len(df))
print("Rows after duplicate removal:", len(df_clean))
print("Duplicate rows removed:", len(df) - len(df_clean))
print("Remaining duplicate rows:", df_clean.duplicated().sum())
```

```python
for column in df_clean.columns:
    placeholders = df_clean[column].astype("string").str.strip().str.lower().isin(
        ["", "none", "null", "nan", "-"]
    )

    print(f"{column}: {placeholders.sum()} placeholder values")
```

```python
print("Actual missing values:")
print(df_clean.isna().sum())

print("\nDate placeholders:")
print(
    df_clean["signup_date"]
    .astype("string")
    .str.strip()
    .str.lower()
    .value_counts()
    .reindex(["none", "-", "-999"], fill_value=0)
)

print("\nSpending placeholders:")
print(
    df_clean["spend"]
    .astype("string")
    .str.strip()
    .str.lower()
    .value_counts()
    .reindex(["none", "-", "-999"], fill_value=0)
)
```

```python
df_clean["signup_date"] = df_clean["signup_date"].replace(
    ["none", "-", "-999"], np.nan, regex=False
)

df_clean["spend"] = df_clean["spend"].replace(
    ["none", "-", "-999"], np.nan, regex=False
)

print("Missing values after placeholder cleaning:")
print(df_clean.isna().sum())
```

```python
# Convert spending values into a consistent numeric format
spend_text = (
    df_clean["spend"]
    .astype("string")
    .str.strip()
    .str.replace(r"(?i)USD\s*", "", regex=True)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df_clean["spend"] = pd.to_numeric(spend_text, errors="coerce")

print("Spending data type:", df_clean["spend"].dtype)
print("\nSpending summary:")
print(df_clean["spend"].describe())
```

```python
# Calculate IQR boundaries
Q1 = df_clean["spend"].quantile(0.25)
Q3 = df_clean["spend"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df_clean[
    (df_clean["spend"] < lower_bound) |
    (df_clean["spend"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))
print("\nLargest spending values:")
print(outliers["spend"].nlargest(10).to_list())
```

```python
# Record outliers before capping
outlier_count_before = len(outliers)

# Cap unusually high spending values at the IQR upper bound
df_clean["spend"] = df_clean["spend"].clip(upper=upper_bound)

print("Outliers identified before capping:", outlier_count_before)
print("Upper capping limit:", round(upper_bound, 2))
print("Maximum spending after capping:", df_clean["spend"].max())
print("Missing spending values:", df_clean["spend"].isna().sum())
```

```python
# Calculate the median of the cleaned spending values
spend_median = df_clean["spend"].median()

# Fill missing spending values with the median
df_clean["spend"] = df_clean["spend"].fillna(spend_median)

print("Median used for imputation:", round(spend_median, 2))
print("Missing spending values remaining:", df_clean["spend"].isna().sum())
print("Spending data type:", df_clean["spend"].dtype)
```

```python
# Standardise country names
df_clean["country"] = df_clean["country"].astype("string").str.strip().str.lower()

country_mapping = {
    "us": "United States",
    "u.s.": "United States",
    "usa": "United States",
    "united states": "United States",
    "uk": "United Kingdom",
    "gb": "United Kingdom",
    "united kingdom": "United Kingdom",
    "fr": "France",
    "france": "France",
    "de": "Germany",
    "germany": "Germany",
    "br": "Brazil",
    "brazil": "Brazil"
}

df_clean["country"] = df_clean["country"].replace(country_mapping)

print(df_clean["country"].value_counts())
```

```python
# Standardise customer segment labels
df_clean["segment"] = (
    df_clean["segment"]
    .astype("string")
    .str.strip()
    .str.lower()
)

segment_mapping = {
    "smb": "Small Business",
    "small business": "Small Business",
    "consumer": "Consumer",
    "ent": "Enterprise",
    "enterprise": "Enterprise",
    "midmarket": "Mid-Market",
    "mid-market": "Mid-Market",
    "mid market": "Mid-Market"
}

df_clean["segment"] = df_clean["segment"].replace(segment_mapping)

print(df_clean["segment"].value_counts())
```

```python
# Standardise active status
active_text = (
    df_clean["is_active"]
    .astype("string")
    .str.strip()
    .str.lower()
)

active_mapping = {
    "true": True,
    "1": True,
    "yes": True,
    "y": True,
    "false": False,
    "0": False,
    "no": False,
    "n": False
}

df_clean["is_active"] = active_text.map(active_mapping).astype("boolean")

print("Active status counts:")
print(df_clean["is_active"].value_counts(dropna=False))
print("\nData type:", df_clean["is_active"].dtype)
```

```python
# Standardise customer names
df_clean["full_name"] = (
    df_clean["full_name"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Standardise email addresses
df_clean["email"] = (
    df_clean["email"]
    .astype("string")
    .str.strip()
    .str.lower()
)

print("Sample standardised names:")
print(df_clean["full_name"].head(10))

print("\nSample standardised emails:")
print(df_clean["email"].head(10))
```

```python
# Convert mixed date formats into datetime
df_clean["signup_date"] = pd.to_datetime(
    df_clean["signup_date"],
    format="mixed",
    errors="coerce"
)

print("Signup date data type:", df_clean["signup_date"].dtype)
print("\nEarliest signup date:", df_clean["signup_date"].min())
print("Latest signup date:", df_clean["signup_date"].max())
print("Missing signup dates:", df_clean["signup_date"].isna().sum())
```

```python
# Standardise customer IDs
df_clean["customer_id"] = (
    df_clean["customer_id"]
    .astype("string")
    .str.strip()
)

print("Customer ID data type:", df_clean["customer_id"].dtype)
print("\nSample customer IDs:")
print(df_clean["customer_id"].head(10))
```

```python
# Find customer IDs that appear more than once
repeated_ids = df_clean["customer_id"].value_counts()
repeated_ids = repeated_ids[repeated_ids > 1]

print("Number of customer IDs appearing more than once:", len(repeated_ids))
print("\nMost frequently repeated IDs:")
print(repeated_ids.head(10))

print("\nExample records for repeated IDs:")
print(
    df_clean[df_clean["customer_id"].isin(repeated_ids.index)]
    [["customer_id", "full_name", "email", "signup_date"]]
    .head(15)
)
```

```python
# Check for exact duplicate rows after standardisation
duplicate_rows = df_clean[df_clean.duplicated(keep=False)]

print("Rows involved in exact duplicates:", len(duplicate_rows))
print("\nExact duplicate records:")
print(duplicate_rows.sort_values("customer_id").head(10))
```

```python
# Remove exact duplicates revealed by standardisation
rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates().copy()

rows_after = len(df_clean)

print("Rows before removal:", rows_before)
print("Rows after removal:", rows_after)
print("Additional duplicate rows removed:", rows_before - rows_after)
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```

### Data Cleaning Decisions

1. **Exact duplicates:** Removed exact duplicate records after standardising the data. The initial dataset contained 1,000 rows. One exact duplicate was removed initially, and 42 additional duplicate rows became identifiable after standardisation, leaving 957 rows.

2. **Missing email addresses:** Retained missing values because email addresses cannot be reliably inferred from other fields.

3. **Missing signup dates:** Retained missing values because the original dates are unknown and should not be invented.

4. **Missing spending values:** Converted placeholders (`none`, `-`, and `-999`) to missing values, then filled missing spending with the median of the valid, outlier-capped spending values.

5. **Spending outliers:** Detected outliers using the IQR method. Values above the upper IQR boundary were capped at approximately 1,381.03 to reduce the influence of extreme observations without deleting customer records. This is a statistical treatment, not proof that every flagged value was erroneous.

6. **Country names:** Standardised country variants into United States, United Kingdom, Germany, Brazil, and France.

7. **Customer segments:** Standardised segment labels into Consumer, Small Business, Enterprise, and Mid-Market.

8. **Active status:** Converted variations such as `Y`, `yes`, `1`, and `true` into Boolean `True`; variations such as `N`, `no`, `0`, and `false` into Boolean `False`.

9. **Names and emails:** Trimmed extra spaces, converted names to title case, and converted email addresses to lowercase. Email validity still requires separate validation.

10. **Data types:** Converted customer IDs to strings, signup dates to datetime, spending to numeric, and active status to Boolean.

```python
# Identify emails that do not match a basic email format
email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

has_email = df_clean["email"].notna()

invalid_emails = df_clean.loc[
    has_email & ~df_clean["email"].str.match(email_pattern, na=False),
    ["customer_id", "full_name", "email"]
]

print("Non-missing emails checked:", has_email.sum())
print("Invalid email formats found:", len(invalid_emails))
print("\nExamples of invalid emails:")
print(invalid_emails.head(15))
```

```python
# Extract clearly identifiable email addresses inside angle brackets
wrapped_email = df_clean["email"].str.extract(r"<([^<>]+)>", expand=False)

df_clean["email"] = df_clean["email"].where(
    wrapped_email.isna(),
    wrapped_email
)

# Recheck email formats
email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

has_email = df_clean["email"].notna()

invalid_emails = df_clean.loc[
    has_email & ~df_clean["email"].str.match(email_pattern, na=False),
    ["customer_id", "full_name", "email"]
]

print("Remaining invalid email formats:", len(invalid_emails))
print("\nExamples:")
print(invalid_emails.head(10))
```

```python
# Mark unresolved invalid email formats as missing
email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

has_email = df_clean["email"].notna()

invalid_mask = has_email & ~df_clean["email"].str.match(
    email_pattern, na=False
)

invalid_count = int(invalid_mask.sum())

df_clean.loc[invalid_mask, "email"] = pd.NA

print("Invalid email formats marked missing:", invalid_count)
print("Missing emails after cleaning:", df_clean["email"].isna().sum())
print("Invalid non-missing emails remaining:", (
    df_clean["email"].notna()
    & ~df_clean["email"].str.match(email_pattern, na=False)
).sum())
```
